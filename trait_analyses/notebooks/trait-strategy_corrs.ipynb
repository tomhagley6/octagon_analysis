### Pseudonym lists

```python
import re
pseudo_tuples_list = []
experiment_dates_socials = []
for file in social_files:
     # match the session number and the pseudonym string
    match = re.search(r'(\d+_\d)[/\\](\d{4}-\d{2}-\d{2})_(\d{2}-\d{2}-\d{2})_(.*?_.*?)_Social\.json', file)
    if not match:
        print(f"Filename does not match expected pattern: {file}")
    if match:
        session, date, time, pseudonyms = match.groups()
        pseudo_tuples = pseudonyms.split('_')[0::]
        pseudo_tuples_list.append(pseudo_tuples)
        experiment_dates_socials.append((date, pseudo_tuples))
```

```python
experiment_dates_socials
```

```python
first_solo_pseudo_list = []
second_solo_pseudo_list = []
first_solo_dates = []
second_solo_dates = []
solo_dates = {}


for first_solo, second_solo in zip(first_solos, second_solos):
   # match the session number and the pseudonym string
    first_match = re.search(r'(\d+_\d)[/\\](\d{4}-\d{2}-\d{2})_(\d{2}-\d{2}-\d{2})_(.*?)_FirstSolo\.json', first_solo)
    second_match = re.search(r'(\d+_\d)[/\\](\d{4}-\d{2}-\d{2})_(\d{2}-\d{2}-\d{2})_(.*?)_SecondSolo\.json', second_solo)

    if not first_match:
        print(f"Filename does not match expected pattern: {first_solo}")
    if first_match:
        session, date, time, pseudonyms = first_match.groups()
        first_pseudo = pseudonyms.split('_')[0::]
        first_solo_pseudo_list.append(first_pseudo)
        first_solo_dates.append(date)

    if not second_match:
        print(f"Filename does not match expected pattern: {second_solo}")
    if second_match:
        session, date, time, pseudonyms = second_match.groups()
        second_pseudo = pseudonyms.split('_')[0::]
        second_solo_pseudo_list.append(second_pseudo)
        second_solo_dates.append(date)

for i, pseudonym in enumerate(first_solo_pseudo_list):
    player = pseudonym[0]
    if player not in solo_dates:
        solo_dates[player] = []
    solo_dates[player] = first_solo_dates[i]
```

## Social

```python
# trial logic
scores_per_unit_time = [(0, 0) for _ in range(len(social_lists))]
delta_times_list = []
for trial_list_index in range(len(social_lists)):
    # define trial
    trial_list = social_lists[trial_list_index]
    delta_times = []
    for trial_index in range(len(trial_list)):
        trial = trial_list[trial_index]
        # cut trials to only return timepoints between slice onset and server-selected trigger activation
        slice_onset = trial[trial['eventDescription'] == globals.SLICE_ONSET]
        slice_onset_index = slice_onset.index[0] - trial.index[0]
        selected_trigger_activation = trial[trial['eventDescription'] == globals.SELECTED_TRIGGER_ACTIVATION]
        selected_trigger_activation_index = selected_trigger_activation.index[0] - trial.index[0]

        # extract application time between slice onset and server-selected trigger activation
        time_app_onset = trial['timeApplication'].iloc[slice_onset_index]
        time_app_trigger = trial['timeApplication'].iloc[selected_trigger_activation_index]        
        time = time_app_trigger - time_app_onset
        delta_time = time.total_seconds()
        delta_times.append(delta_time)
        
    total_time = sum(delta_times)
    delta_times_list.append(total_time)

    # access final trial event log event for the final player scores
    final_trial = trial_list[-1]
    final_trial_trial_end = final_trial[final_trial['eventDescription'] == 'trial end']
    
    # flexibly index player scores
    player0_score = final_trial_trial_end[globals.PLAYER_SCORE_DICT[0]['score']].item()
    player1_score = final_trial_trial_end[globals.PLAYER_SCORE_DICT[1]['score']].item()

    # score per unit time as final score divided by total time spent playing (active trials)
    score_per_time_unit_p0 = player0_score/total_time
    score_per_time_unit_p1 = player1_score/total_time
    print(player0_score, player1_score)
    print(score_per_time_unit_p0, score_per_time_unit_p1)
    scores_per_unit_time[trial_list_index] = (score_per_time_unit_p0, score_per_time_unit_p1)
```

```python
scores_per_unit_time_dict = {}

for i, tuple in enumerate(pseudo_tuples_list):
    player0 = tuple[0]
    player1 = tuple[1]

    if player0 not in scores_per_unit_time_dict:
        scores_per_unit_time_dict[player0] = []
    if player1 not in scores_per_unit_time_dict:
        scores_per_unit_time_dict[player1] = []
    
    scores_per_unit_time_dict[player0] = scores_per_unit_time[i][0]
    scores_per_unit_time_dict[player1] = scores_per_unit_time[i][1]
```

```python
# trial logic
scores_per_unit_time_comb_solos = [0 for _ in range(len(combined_solos))]
delta_times_list_comb_solos = []
for trial_list_index in range(len(combined_solos)):
    # define trial
    trial_list = combined_solos[trial_list_index]
    delta_times = []
    for trial_index in range(len(trial_list)):
        trial = trial_list[trial_index]
        # cut trials to only return timepoints between slice onset and server-selected trigger activation
        slice_onset = trial[trial['eventDescription'] == globals.SLICE_ONSET]
        slice_onset_index = slice_onset.index[0] - trial.index[0]
        #selected_trigger_activation = trial[trial['eventDescription'] == globals.SELECTED_TRIGGER_ACTIVATION]

        selected_trigger_activation = trial[
            trial['eventDescription'].astype(str).str.strip()
            == str(globals.SELECTED_TRIGGER_ACTIVATION).strip()
        ]

        if selected_trigger_activation.empty:
            print(trial_list_index, trial_index)
            print("Skipping trial: no selected trigger activation found")
            continue

        selected_trigger_activation_index = selected_trigger_activation.index[0] - trial.index[0]

        # extract application time between slice onset and server-selected trigger activation
        time_app_onset = trial['timeApplication'].iloc[slice_onset_index]
        time_app_trigger = trial['timeApplication'].iloc[selected_trigger_activation_index]        
        time = time_app_trigger - time_app_onset
        delta_time = time.total_seconds()
        delta_times.append(delta_time)
        
    total_time = sum(delta_times)
    delta_times_list.append(total_time)

    # access final trial event log event for the final player scores
    final_trial = trial_list[-1]
    final_trial_trial_end = final_trial[final_trial['eventDescription'] == 'trial end']
    
    # flexibly index player scores
    player0_score = final_trial_trial_end[globals.PLAYER_SCORE_DICT[0]['score']].item()
    # score per unit time as final score divided by total time spent playing (active trials)
    score_per_time_unit_p0 = player0_score/total_time

    scores_per_unit_time_comb_solos[trial_list_index] = score_per_time_unit_p0
```

```python
for target in targets:

    print(f"\n{'='*60}")
    print(f"Regression results for {target}")
    print(f"{'='*60}")

    for predictor in predictors:

        # remove NaNs pairwise
        subset = df[[predictor, target]].dropna()

        # skip tiny datasets
        if len(subset) < 3:
            continue

        x = subset[predictor]
        y = subset[target]

        result = linregress(x, y)

        print(
            f"{predictor:40s} "
            f"r={result.rvalue: .3f} "
            f"r²={result.rvalue**2: .3f} "
            f"p={result.pvalue: .4f}"
        )

        # plot
        fig, ax = plt.subplots(figsize=(5,4))

        # scatter points
        ax.scatter(
            x,
            y,
            alpha=0.65,
            s=80,
            color='steelblue'
        )

        # regression line
        xfit = np.linspace(x.min(), x.max(), 100)
        yfit = result.slope * xfit + result.intercept

        ax.plot(
            xfit,
            yfit,
            linewidth=3,
            color='darkorange'
        )

        # labels
        ax.set_xlabel(predictor, fontsize=20)
        ax.set_ylabel(target, fontsize=20)

        ax.set_title(
            f"{target} vs {predictor}\n"
            f"r={result.rvalue:.2f}, p={result.pvalue:.3f}",
            fontsize=20
        )

        # remove top and right spines
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

        # tick font sizes
        ax.tick_params(axis='both', labelsize=20)

        plt.tight_layout()
        plt.show()
```
